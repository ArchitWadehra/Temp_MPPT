```python
import gym
import gym_mppt
from gym import envs

env = gym.make("mppt_shaded-v2")

num_states = env.observation_space.shape[0]
print("Size of State Space ->  {}".format(num_states))
print(env.observation_space)

print()

num_actions = env.action_space.shape[0]
print("Size of Action Space ->  {}".format(num_actions))
print(env.action_space)

upper_bound = env.action_space.high[0]
lower_bound = env.action_space.low[0]

print("Max Value of Action ->  {}".format(upper_bound))
print("Min Value of Action ->  {}".format(lower_bound))
```

```python
env.reset()
powerlst = []
rewlst = []

totalrew = []
# print(env.buf_infos)

# for i in range(300):
#     obs, rewards, dones, info = env.step(-1)
#     powerlst.append(obs["Power"][0])
#     print(obs)
#     print(info)
#     print(rewards)
#     print()
for i in range(212):
    obs, rewards, dones, info = env.step(1)
    powerlst.append(obs[1])
    rewlst.append(rewards)
    
for i in range(24):
    obs, rewards, dones, info = env.step(-5)
    powerlst.append(obs[1])
    rewlst.append(rewards)

for i in range(100 - 40 - 22):
    obs, rewards, dones, info = env.step(0)
    powerlst.append(obs[1])
    rewlst.append(rewards)

import matplotlib.pyplot as plt
from matplotlib.pyplot import figure

figure(figsize=(10, 4), dpi=500)
# plt.plot(powerlst)
plt.plot(rewlst)
```
