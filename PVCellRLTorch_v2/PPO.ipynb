```python
from stable_baselines3 import PPO
from stable_baselines3.common.env_util import make_vec_env
```

```python
# Parallel environments
env = make_vec_env("mppt_shaded-v1", n_envs=4)

model = PPO("MlpPolicy", env, verbose=1, learning_rate=0.000001)
model.learn(total_timesteps=40000000, progress_bar=True)
model.save("PPO_v1")
```

```python
model = PPO.load("ppo_cartpole")

obs = env.reset()
while True:
    action, _states = model.predict(obs)
    obs, rewards, dones, info = env.step(action)
    env.render()
```
