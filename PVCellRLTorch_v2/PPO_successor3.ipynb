```python
from stable_baselines3 import PPO
from stable_baselines3.common.env_util import make_vec_env
```

```python
# Parallel environments
env = make_vec_env("mppt_shaded-v4", n_envs=64)
TIMESTEPS = 10000000
# TIMESTEPS = 100001
tensorboard_log = os.path.join(logdir, "tensorboard")

model = PPO("MlpPolicy", env, verbose=1, learning_rate=0.001, tensorboard_log=tensorboard_log)
model.learn(total_timesteps=TIMESTEPS, progress_bar=True, reset_num_timesteps=False, tb_log_name="PPO-successor6")
# model.save(f"{models_dir}/{TIMESTEPS}")
env = model.get_env()
```

```python
for i in range(10):
    obs = env.reset()
    print(env.buf_infos)
    for j in range(100):
        action, _states = model.predict(obs)
        obs, rewards, dones, info = env.step(action)
        print(obs)
        print(rewards)
    print()   
```
