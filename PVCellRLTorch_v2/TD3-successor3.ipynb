```python
import gym
import gym_mppt
from gym import envs

import os
models_dir = "models/TD3"
logdir = "logs"

if not os.path.exists(models_dir):
    os.makedirs(models_dir)
    
if not os.path.exists(logdir):
    os.makedirs(logdir)    
```

```python
env = gym.make("mppt_shaded-v4")

num_actions = env.observation_space.shape[0]
print("Size of State Space ->  {}".format(num_actions))
print(env.observation_space)

# num_states = len(env.observation_space)
# print("Size of State Space ->  {}".format(num_states))
# print(env.observation_space)

print()

num_actions = env.action_space.shape[0]
print("Size of Action Space ->  {}".format(num_actions))
print(env.action_space)

upper_bound = env.action_space.high[0]
lower_bound = env.action_space.low[0]

# print("Max Value of Action ->  {}".format(upper_bound))
# print("Min Value of Action ->  {}".format(lower_bound))
```

```python
import numpy as np
from stable_baselines3 import TD3
from stable_baselines3.common.noise import NormalActionNoise, OrnsteinUhlenbeckActionNoise
```

```python
n_actions = env.action_space.shape[-1]
action_noise = NormalActionNoise(mean=np.zeros(n_actions), sigma=0.0 * np.ones(n_actions))

tensorboard_log = os.path.join(logdir, "tensorboard")
model = TD3("MlpPolicy", env, action_noise=action_noise, verbose=1, learning_rate=0.001, tensorboard_log=tensorboard_log)
# TIMESTEPS = 1000002  # 3 hours
TIMESTEPS = 300000
model.learn(total_timesteps=TIMESTEPS, log_interval=10, progress_bar=True, reset_num_timesteps=False, tb_log_name="TD3-successor6")
# model.save(f"{models_dir}/{TIMESTEPS}")
env = model.get_env()
```

```python
import matplotlib.pyplot as plt
powerlst = []
voltlst = []
for i in range(1):
    obs = env.reset()
    print(env.buf_infos)
    for j in range(100):
        action, _states = model.predict(obs)
        obs, rewards, dones, info = env.step(action)
        print(obs)
        print(rewards)
#         print(obs['Power'][0])
        powerlst.append(obs['Power'][0])
        voltlst.append(obs['Voltage'][0])
#         print(rewards)
    plt.plot(powerlst)
    print()   
```
